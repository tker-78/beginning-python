## 自分なりの実装をしてみる
### 実装の方針
- 見出し部分には<h2>タグをつける
- 見出しかどうかの判定を行う
- リストには`・`をつける
- リストかどうかの判定を行う
- 80文字以上の連続文字はパラグラフとして判定する
- パラグラフには`<p>`タグをつける
- メールアドレスかどうかの判定を行う
- メールアドレスには`mailto:`をつける
- webのリンクかどうかの判定を行う
- webのリンクには、`<a href="...">`をつける
- `*`はboldと判定する


## 構成
下記の構成にする
- 判定を行うプログラム: `parser,py`
- マークアップを行うルールを規定するプログラム: `rule.py`
- マークアップを実行するプログラム: `markup.py`

### Handlerクラス
Parserで解析した結果から、付与するマークアップの種類に基づき、Handlerに情報を渡す。
その情報に基づき、Handlerは処理を実行する。
Handlerが行うのは、適切なメソッドの割り当て。

### Ruleクラス
ここで定義したルールは、Parserの属性に追加することで、Parser毎に適用するルールを柔軟に設定できるようにする。

-
### Parserクラス
Parserは、テキストを解析し、各要素を特定する。

### Filterクラス
Parserで解析した結果をもとに、テキストの変換処理を行う。
Handlerクラス内に定義する。


なので、呼び出される順番は、
- Parser -> Handler -> Rule -> Filter

## 実装メモ

最初に呼び出しのインターフェースを作るため、markup.pyを実装する。
呼び出しの実行部分と、Parser, BasicTextParserの枠組みを書く。
できたら、次にHandlerの実装を行う。

HTMLに関するメソッドは`HTMLRenderer`に、汎用的なメソッドや呼び出すメソッドを振り分けるcallbackメソッドは`Handler`に定義する。

HTMLRendererに定義するメソッド
- start_paragraph
- end_paragraph
- start_heading
- end_heading
- start_list
- end_list
- start_listitems
- end_listitems
- start_title
- end_title

Handlerに定義するメソッド
- callback
- sub
